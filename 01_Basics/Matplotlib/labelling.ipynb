```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Load Data
raw_data = load_dataset("lukebarousse/data_jobs")
df = raw_data['train'].to_pandas()

# Cleanup data
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df.info()
```

##### Adding X and Y axis Labels, rotating Label title names in order to fit properly and align them to Right, Left or Center of the Axis. Applying Dark theme

```python
job_counts = df['job_title_short'].value_counts()
job_counts.sort_values(ascending=False, inplace=True) # Not needed as .value_counts() by default sorts in Descending order of Index values
print(job_counts, end = "\n\n\n")

x = job_counts.index
y = job_counts


plt.style.use("dark_background")
plt.bar(x, y, color = "#A2FF00")
plt.title("Plot of Job Postings Count w.r.t every month\n")
plt.xlabel("Job Title")
plt.ylabel("Job Postings Count")
plt.xticks(rotation = 45, ha = "right", fontsize = 7)
plt.show()
```
