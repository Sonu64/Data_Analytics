```python
import pandas as pd
import matplotlib.pyplot as plt

# Setup: A Series of average salaries by month
s = pd.Series([95000, 105000, 110000], index=[1, 2, 3], name='salary')


# --- MATPLOTLIB WAY ---
# You MUST pass the data arrays explicitly.
plt.plot(s.index, s.values, marker='o')
plt.title("Matplotlib Series Plot")
plt.xlabel("INDEX values")
plt.ylabel("Values corresponding to those X values")
plt.show()

# --- PANDAS WAY ---
# While Series auto-maps Index -> X and Values -> Y, 
# we can explicitly pass the data arrays to x and y.
# Note: In a Series, y can be the 'name' string OR .values, but x must be the data itself, i.e-> .index
# Best Strategy for Series if using plot() with pandas is not to use either x and y, just put the labels according to the Chart type (Vertical OR Horizontal)
s.plot(kind='bar', x=s.index, y=s.values)
plt.xlabel("INDEX values")
plt.ylabel("Values corresponding to those X values")
plt.show()
```
