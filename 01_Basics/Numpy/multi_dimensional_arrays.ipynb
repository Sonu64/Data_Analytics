```python
three_d_list_alphabets = [
    # Layer 1 (Depth = 0)
    [
        # Row 1
        ['A', 'B', 'X'],
        # Row 2
        ['C', 'D', 'X']
    ],
    # Layer 2 (Depth = 1)
    [
        # Row 1
        ['E', 'F', 'X'],
        # Row 2
        ['G', 'H', 'X']
    ],
    # Layer 3 (Depth = 2)
    [
        # Row 1
        ['I', 'J', 'X'],
        # Row 2
        ['K', 'L', 'X']
    ]
]

alphabets = np.array(three_d_list_alphabets)
print(alphabets.ndim)
print(alphabets.shape) # Layers, Rows, Columns
```

```python
name = alphabets[0,0,0] + alphabets[1, 0, 1] + alphabets[2, 0, 0] + alphabets[2, 1, 0]
print(name) # AFIK
```
