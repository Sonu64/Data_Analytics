```python
import numpy as np, random 

salary = np.array([random.randint(1, 101) for num in range(10)])

# Cumulative Sum and Cumulative Product operations
cum_sum_salaries = np.cumsum(salary)
cum_prod_salaries = np.cumprod(salary)

print(salary)
print(cum_sum_salaries)
print(cum_prod_salaries)
```

```python
from numpy.random import Generator as gen 
from numpy.random import PCG64 as pcg 

arr_random_generator = gen(pcg(seed = 42)) # We tell NumPy to create a Random generator object which will generate random numbers using PCG algorithm
random_array = arr_random_generator.normal(size=(5, 5)) # Random numbers from -1 to 1 excluded, following the Bell Curve
print(random_array)
```

```python
# Random numbers between 0 and 1 excluded, totally Random, not following any Curve
arr_random_generator = gen(pcg(seed = 10))
random_nums = arr_random_generator.random(size = (3, 3))
print(random_nums)
```

```python
# Random Integers between a LOW and HIGH range, HIGH excluded
arr_random_generator = gen(pcg(seed = 41))
randomInts = arr_random_generator.integers(low = 1, high = 101, size = (3, 3))
print(randomInts)
```

```python
# Random Floating Point Numbers using .uniform(), for Uniform Distribution

arr_random_generator = gen(pcg(seed = 90))
random_floats = arr_random_generator.uniform(low = 1, high = 21, size = (2, 3))
print(random_floats)
```

```python
arr_random_generator = gen(pcg(seed = 100))
arr = np.array([1, 2, 3, 4, 5])
arr_random_generator.shuffle(arr)
print(arr)
```
