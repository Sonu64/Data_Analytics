## Broadcasting in Numpy

![BROADCASTING](broadcasting.png)

```python
import numpy as np

# For a successful Broadcast w.r.t a Row or Column, the dimension of Rows should either be same or one of them should be 1 (this will be repeated). Same rule applies to Columns as well.
arr1 = np.array([[1, 2, 3, 4]]) # 1 Row, 4 Columns
arr2 = np.array([[1], [2], [3], [4]]) # 4 Rows, 1 Column
print(arr1.shape, arr2.shape)
```
