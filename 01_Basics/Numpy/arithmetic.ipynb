```python
import numpy as np

# Scalar Arithmetic -> One input is scalar, like a number to be added.
# Under the hood this is also Vector Arithmetic as 1 is converted to [1, 1, 1] and added vectorially in the array, ElementWise. But here one Input is Scalar, so can be thought as Scalar Arithmetic

array = np.array([1, 2, 3])
print (array + 1)
print (array - 2)
print(array * 3)
print (array/ 4)
print (array ** 5)
```

```python
# Vector Arithmetic -> Element wise operation -> Input is a Single NxN matrix -> A vector
print(np.sqrt(array))
print(np.round(array))
print(np.floor(array))
print(np.ceil(array))
```

```python
radii = np.array([1, 2, 3])
areas = radii ** 2 * np.pi 
print(areas)
```

```python
# Element wise operations on 2 or more arrays -> More 'Vector Like' where inputs are NxN matrices
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)
print(arr1 ** arr2)
```

```python
# Creating Boolean Arrays using Conditionals
scores = np.array([60, 62, 90, 100, 45, 32, 100])

# Students who passed
print(scores >= 60)
# Students who failed
print(scores < 60)
# Students with Full marks
print(scores == 100)
```

```python
# Assigning values to Truthy values in filtered array
scores[scores < 60] = 0
print(scores)
```
