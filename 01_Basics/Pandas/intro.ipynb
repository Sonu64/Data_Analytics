```python
# Using Hugging Face to get a sample large dataset and fetch it via load_dataset
# For load_datasets to work we first install the datasets library provided by Hugging Face
from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs") # Type is DatasetDict
print(type(ds)) # DatasetDict
training_data = ds['train'] # Get the training data from the dataset
df = training_data.to_pandas() # CONVERT TO A PANDAS DATAFRAME
print(type(df)) # DataFrame
print(df)
```
