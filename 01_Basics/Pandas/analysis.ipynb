```python
"""Applying Aggregate Functions on Individual Columns"""

# Loading Data
import pandas as pd
from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs") # Type is DatasetDict
training_data = ds['train'] # Get the training data from the dataset
df = training_data.to_pandas() # CONVERT TO A PANDAS DATAFRAME

# Convert job_posted_date column to pandas datetime object --> Cleaning
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)

# Single Numeric Column
x = df['salary_year_avg'].mean()
print(x)
# Multiple Numeric Columns by passing a List of Column Names
df[['salary_year_avg', 'salary_hour_avg']].mean()



print("\n\n\n\n\n_____________________________________\n\n\n\n")
# Get the Minimum Yearly Salary, then get its index, then use iloc to print details about that Row
min_salary = df['salary_year_avg'].min()
min_salary_index = df['salary_year_avg'].idxmin() # Get index of min value
print(df.iloc[min_salary_index])
# Print only the 1st 2 columns of that row using iloc
print("\n\n\n___________________\n\n\n")
print(df.iloc[min_salary_index, 0:2])





# Get how many times a value of a Column occurs in all the Rows
# See all Unique Values
print("\n\n\n" + str(df['job_title_short'].unique()) + "\n\n\n")
# See how many times those values occur individually
print(df['job_title_short'].value_counts())
```

```python
"""Using the GROUP_BY aggregate function"""
# Loading Data
import pandas as pd
from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs") # Type is DatasetDict
training_data = ds['train'] # Get the training data from the dataset
df = training_data.to_pandas() # CONVERT TO A PANDAS DATAFRAME

# Convert job_posted_date column to pandas datetime object --> Cleaning
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)


# Create 10 different groups for 10 different job_title_short
print(df.groupby('job_title_short'))
# After group by we generally use a Aggregate function to apply on a particular column in the groups
print(df.groupby('job_title_short')['salary_year_avg'].min()) # 10 groups, 10 min values for salary_year_avg
print("\n\n\n")
max_salaries_in_countries = (df.groupby('job_country').salary_year_avg.max()) # A Series
print(max_salaries_in_countries['India']) # Access values in a Series by KeyName directly !


print("\n\n\n\n")



# Apply the Aggregate Functions to Multiple Columns
df.groupby('job_title_short')[['salary_year_avg', 'salary_hour_avg']].mean()
# Apply multiple Aggregate Functions to those Column(s), by passing a list of function names to .agg([])
df.groupby('job_title_short')[['salary_year_avg', 'salary_hour_avg']].agg(['min', 'max', 'mean', 'median'])


# Group by Multiple Columns
df.groupby(['job_title_short', 'job_country'])['salary_year_avg'].min()
```
