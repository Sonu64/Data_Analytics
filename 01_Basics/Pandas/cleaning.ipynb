```python
from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs") # Type is DatasetDict
training_data = ds['train'] # Get the training data from the dataset
df = training_data.to_pandas() # CONVERT TO A PANDAS DATAFRAME

# Convert job_posted_date column to pandas datetime object
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)

# Once a Column is of type datetime(64), the .dt property can be accessed on the values to select specific parts of the date like month, year, day, etc....
print("DD: " + str(df['job_posted_date'].dt.day))
print("MM: " + str(df['job_posted_date'].dt.month))
print("YYYY: " + str(df['job_posted_date'].dt.year))


# Create a new Column in the Dataframe object with that new data received, Only Supports SUBSCRIPT notation !!!!!
df['job_posted_month'] = df['job_posted_date'].dt.month;

# Show updated Dataframe
df.info()
```

```python
"""SORTING ROWS BY THE VALUES OF A PARTICULAR COLUMN"""

from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs") # Type is DatasetDict
training_data = ds['train'] # Get the training data from the dataset
df = training_data.to_pandas() # CONVERT TO A PANDAS DATAFRAME

# Convert job_posted_date column to pandas datetime object
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)


# Create a new Column in the Dataframe object with that new data received, Only Supports SUBSCRIPT notation !!!!!
df['job_posted_month'] = df['job_posted_date'].dt.month;
# df.sort_values(by = column_name, inplace = True or False based on wheather in-place sorting required or not)
# Not in-place ---> Reassigning to original df
# df = df.sort_values(by = "job_posted_month", inplace = False), below we are printing only the job_posted_month column of those Rows
print(df.sort_values(by = "job_posted_month", inplace = False)['job_posted_month'], end = "\n\n\n____________________________________________\n\n\n")


# Create a new column called job_posted_year
df['job_posted_year'] = df['job_posted_date'].dt.year
df.info()
# In-place Sort
df.sort_values(by = 'job_posted_year', inplace = True)
# The select only the 'job_posted_year' column from those rows to check for Sort
print(df.job_posted_year.tail(30))
```
