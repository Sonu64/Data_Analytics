```python
from datasets import load_dataset

ds = load_dataset("lukebarousse/data_jobs")
training_data = ds['train']
df = training_data.to_pandas()

# Printing TOP n Rows or LAST n ROWS using .head(<n>) or .tail(<n>)
print(df.head()) # Top 5
print(df.head(10), end = "\n\n") # Top 10
print(df.tail(), end = "\n\n") # Last 5
print(df.tail(10)) # Last 10 
```

```python
# Choose specific columns and print accordingly
from datasets import load_dataset

ds = load_dataset("lukebarousse/data_jobs")
training_data = ds['train']
df = training_data.to_pandas()

print(df.job_title_short.head(3), end = "\n\n") # Top 3 values of the job_title_short column

print(df[['job_title_short', 'job_title', 'job_location']].head(3)) # Selecting multiple columns by providing a List of Column names to df subscript

# Select a Row using index -> Can't be Done Normally --> ERROR !!!!!
print((df[['job_title_short', 'job_title', 'job_location']].head(3))[1])
```

```python
# Print Rows using iloc
from datasets import load_dataset

ds = load_dataset("lukebarousse/data_jobs")
training_data = ds['train']
df = training_data.to_pandas()

# Select a single row using index
print(df.iloc[4], end = "\n\n\n\n")

# Select multiple rows using slicing
print(df.iloc[0:3], end = "\n\n\n") # Top 2 Rows

# Select multiple Rows 1st -> Then select specific Sliced columns from that Row
print(df.iloc[0:3, 0:2], end = "\n\n\n")

# Select multiple Columns using Subscript notation and then select specific Rows from it
print(df[['job_title_short', 'job_title', 'job_location']].head(20).iloc[9], end = "\n\n")
print(df[['job_title_short', 'job_title', 'job_location']].head(10).iloc[9])
print(df.iloc[9])
```
