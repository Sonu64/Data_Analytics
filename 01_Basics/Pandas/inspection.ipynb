```python
# Getting Rows with One or multiple conditions for the column values

from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs") # Type is DatasetDict
training_data = ds['train'] # Get the training data from the dataset
df = training_data.to_pandas() # CONVERT TO A PANDAS DATAFRAME

print(df[ (df['job_title_short'] == 'Software Engineer') & (df['salary_year_avg']>100000) ] )
print(df['job_title_short'])
```

```python
# df.notna() - displays if the values are not NA (missing values).
# It returns a Boolean same-sized object indicating if the values are not NA.
# Non-missing values are True
# Missing values are False

from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs") # Type is DatasetDict
training_data = ds['train'] # Get the training data from the dataset
df = training_data.to_pandas() # CONVERT TO A PANDAS DATAFRAME

print(df['salary_year_avg'].notna(), end = "\n\n\n") # Boolean Table
# Combine conditions
# Print only those rows where either of Yearly Salary Avg and Hourly Salary Avg are not NA.
print(df[ df['salary_year_avg'].notna() | df['salary_hour_avg'].notna() ], end = "\n\n\n")

# apply this to the entire DataFrame to get a boolean DataFrame indicating where values aren't missing.
print(df.notna())
```
