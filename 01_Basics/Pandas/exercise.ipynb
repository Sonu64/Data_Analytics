```python
# Loading Data
import pandas as pd
from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs") # Type is DatasetDict
training_data = ds['train'] # Get the training data from the dataset
df = training_data.to_pandas() # CONVERT TO A PANDAS DATAFRAME

# Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])

# India based jobs
india_jobs = df[ df['job_country'] == "India" ]
print("Average Yearly Salary in India: ₹" + str(india_jobs['salary_year_avg'].mean()))

# Average and Maximum of yearly avg salaries of each job in India, Sorted from Highest to Lowest according to Avg. salary
jobsData = india_jobs.groupby('job_title_short')['salary_year_avg'].agg(['mean', 'max'])
print("\n\n\nAverage Yearly salary of each job in India - ")
print(jobsData.sort_values(by='mean', ascending=False))

# Sort in place
print("\n\n\n")
india_jobs.sort_values(by = 'salary_year_avg', inplace=True, ascending=False)
print(india_jobs[['job_title_short', 'salary_year_avg']])



# Calculate total Number of Software Engineer jobs in India 
# Calculate count of values in each group using size() and then use 'Software Engineer' as key to index it's count
totalSoftwareJobsIndia = india_jobs.groupby('job_title_short').size()['Software Engineer']
print("Total Software Engineering Jobs in India, calculated using groupby().size(): " + str(totalSoftwareJobsIndia), end = "\n\n")
# Using .value_counts(), and then use 'Software Engineer' as key to index it's count. The Group labels become the index Labels.
totalSoftwareJobsIndia = india_jobs['job_title_short'].value_counts()['Software Engineer'] 
print("Total Software Engineering Jobs in India, calculated using .value_counts : " + str(totalSoftwareJobsIndia), end = "\n\n\n")





# Calculate MIN and MAX salaries for each job
print(india_jobs.groupby(['job_title_short'])[['salary_year_avg']].agg(["min", "max", "count"]), end = "\n\n\n")



# Remove NULL values from the table 
india_jobs_with_Non_Null_AvgYearlySalary = india_jobs.dropna(axis=0, subset=['salary_year_avg'])
india_jobs_with_Non_Null_AvgHourlySalary = india_jobs.dropna(axis=0, subset=['salary_hour_avg'])
# See Non-Null values of each column from .info() on main table
print(india_jobs.info())
# Verify those with the Number of Rows in the below tables
# If Non-Null values for 'salary_year_avg' in main table is X, then rows in below table will be X as well.
print(india_jobs_with_Non_Null_AvgYearlySalary.info(), end = "\n\n")
# If Non-Null values for 'salary_hour_avg' in main table is Y, then rows in below table will be Y as well.
print(india_jobs_with_Non_Null_AvgHourlySalary.info())
```
