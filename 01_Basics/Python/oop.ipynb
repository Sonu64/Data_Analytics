```python
class SonusList:
    def __init__(self):
        """Initialize an empty list."""
        self._items = []

    def add(self, item):
        """Add an item to the end of the list."""
        self._items.append(item)

    def __getitem__(self, index):
        """Retrieve an item by index."""
        return self._items[index]

    def __setitem__(self, index, value):
        """Set an item at a specific index."""
        self._items[index] = value

    def __repr__(self):
        """Return a string representation of the list."""
        return f"Current List wrapped inside object is {self._items}"

    def __len__(self):
        """Return the length of the list."""
        return len(self._items)
```

```python
# Creating an object of type SonusList
ls = SonusList()

#__repr__ dunder will run if object is printed
print(ls)

# add() will run if called on the object
ls.add(34)
ls.add(90)
ls.add(100)

# __setItem__ dunder will run if Square Bracket Notation is used to assign values
# The index should be valid, so have to run .add() beforehand
ls[1] = 111
ls[2] = 1111
print(ls[0], ls[1], ls[2])


# __getItem__ to get the item via [] notation
print(ls[2])

# __len__ will run if len(objName) is called
print(len(ls))
```
