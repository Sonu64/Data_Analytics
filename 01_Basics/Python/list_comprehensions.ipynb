```python
# Using Conditionals 
job_list = [
    "Software Engineer",
    "Data Scientist",
    "Cybersecurity Analyst",
    "Software Engineer",
    "Cloud Architect",
    "Web Developer",
    "Machine Learning Engineer",
    "Data Scientist",
    "DevOps Engineer",
    "Software Engineer"
]
analyst_list = [job for job in job_list if job == "Data Scientist"]
print(analyst_list)
```
