```python
# Calculate total_salary as base_salary * (1 + bonus_rate)
salary_list = [100000, 200000, 150000, 120000, 80000, 750000]
total_salaries = [(lambda x: (x * (1 + 0.1)))(base_salary) for base_salary in salary_list]
print(total_salaries)
```

```python
# Using the filter(function, iterable) function -> Only those items those fulfill the function condition are kept in the iterable, Rest removed
jobs_data = [
    {'job_title': 'Data Scientist',  'job_skills': ['Python', 'Machine Learning'], 'remote': True},
    {'job_title': 'Data Analyst',  'job_skills': ['Excel', 'SQL'], 'remote': False},
    {'job_title': 'Machine Learning Engineer', 'job_skills': ['Python', 'TensorFlow', 'Keras'], 'remote': True},
    {'job_title': 'Software Developer', 'job_skills': ['Java', 'C++'], 'remote': True},
    {'job_title': 'Data Scientist', 'job_skills': ['R', 'Statistics'], 'remote': False}
]

# Define a lambda function for filtering jobs that require Python and are remote
filter_criteria = lambda job: 'Python' in job['job_skills'] and job['remote'] == True

# Jobs requiring Python and are remote
filtered_jobs = filter(filter_criteria, jobs_data) # It will pass each element of jobs_data (named job) to the lambda function filter_criteria(job) and the function will return either TRUE or FALSE based on wheather job satisfies the criteria. If FALSE, job will be removed from jobs_data

print(list(filtered_jobs))
```
