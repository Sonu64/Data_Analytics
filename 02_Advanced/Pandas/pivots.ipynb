### Pivoting in Pandas Dataframes -> To change the Shape of the Dataframe






```python
import pandas as pd
import numpy as np
```

```python
weather_data = [
    {"date": "2025-05-01", "city": "Kolkata", "temp": 34, "humidity": 78},
    {"date": "2025-05-01", "city": "Bankura", "temp": 41, "humidity": 35},
    {"date": "2025-05-01", "city": "Bhubaneswar", "temp": 36, "humidity": 72},
    {"date": "2025-05-01", "city": "Patna", "temp": 38, "humidity": 40},
    {"date": "2025-05-01", "city": "Ranchi", "temp": 32, "humidity": 55},
    {"date": "2025-05-01", "city": "Darjeeling", "temp": 18, "humidity": 85},
    
    {"date": "2025-05-02", "city": "Kolkata", "temp": 35, "humidity": 80},
    {"date": "2025-05-02", "city": "Bankura", "temp": 42, "humidity": 32},
    {"date": "2025-05-02", "city": "Bhubaneswar", "temp": 37, "humidity": 70},
    {"date": "2025-05-02", "city": "Patna", "temp": 39, "humidity": 38},
    {"date": "2025-05-02", "city": "Ranchi", "temp": 33, "humidity": 52},
    {"date": "2025-05-02", "city": "Darjeeling", "temp": 17, "humidity": 88},
    
    {"date": "2025-05-03", "city": "Kolkata", "temp": 33, "humidity": 82},
    {"date": "2025-05-03", "city": "Bankura", "temp": 40, "humidity": 38},
    {"date": "2025-05-03", "city": "Bhubaneswar", "temp": 35, "humidity": 75},
    {"date": "2025-05-03", "city": "Patna", "temp": 40, "humidity": 35},
    {"date": "2025-05-03", "city": "Ranchi", "temp": 31, "humidity": 58},
    {"date": "2025-05-03", "city": "Darjeeling", "temp": 19, "humidity": 82}
]
```

```python
# Convert our data to a Pandas Dataframe
df = pd.DataFrame(weather_data)
# Show top 10 values
df.head(10)
```

```python
# Example-1 : Basic Pivoting to change the Shape of the Dataframe
# For Example, If we wanna create Rows with Unique Date values and Columns with Unique City names
df.pivot(index = 'date', columns = 'city')
# As there are 3 unique dates, 3 Rows will be created for each date
# As there are 6 unique City names, 6 unique Subcolumns will be made for the temp and humidity columns.
# There will be a value for each of temp and humidity for an intersection point of date and city.
```

```python
# The VALUES parameter
# What if we only need one column, let's say I need only temp data in the output --> Use the
# values = [...column names] to specify which column values you need for each intersection point of 
# date and city
df.pivot(index = "date", columns = "city", values = ['temp'])
```
