```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])

# Set index col name to job_index
df.index.name = 'job_index'
```

### 🧡 .reset_index() to reset indices to default integer indices

```python
# Filtering the dataframe for data specific to India only, selecting the job_title_short and salary_year_avg columns

# df.info()

# If we use fillna instead of dropna
median_Indian_salary = df[df['job_country']=="India"][['salary_year_avg']].median()


print("Median Average Yearly Salary in India = Rs. " + str(median_Indian_salary) + ".",  end="\n\n")

indian_salaries = df[df['job_country']=="India"][['job_title_short', 'job_country', 'salary_year_avg']].dropna().sample(10)

indian_salaries.reset_index(inplace=True) 

print(indian_salaries)
# Don't drop the default job_index column yet, as we might need it for merging later

```
