```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])

# Set index col name to job_index
df.index.name = 'job_index'
```

### 🧡 .reset_index() to reset indices to default integer indices

```python
# Filtering the dataframe for data specific to India only, selecting the job_title_short and salary_year_avg columns

# df.info()

# If we use fillna instead of dropna
median_Indian_salary = df[df['job_country']=="India"][['salary_year_avg']].median()
print("Median Average Yearly Salary in India = Rs. " + str(median_Indian_salary) + ".",  end="\n\n")

indian_salaries = df[df['job_country']=="India"][['job_title_short', 'job_country', 'salary_year_avg']].dropna().sample(10, random_state=42)

indian_salaries.reset_index(inplace=True) 

indian_salaries
# Don't drop the default job_index column yet, as we might need it for merging later

```

### 💛 .set_index() to set indices to their initial IDs.

```python
germany_salaries = df[ df['job_country']=='Germany' ][['job_title_short', 'job_country', 'salary_year_avg']].dropna().sample(10, random_state=42)

# First resetting index
germany_salaries.reset_index(inplace=True)
print(germany_salaries, end = "\n\n\n")

# Then removing those indices using set_index
germany_salaries.set_index('job_index', inplace=True)
print(germany_salaries)
```

### 💚 .sort_index() to Sort values according to IDs. Useful in Pivot Tables

```python
# Revise sort_values, it takes a column name(s) in the 'by' parameter and sorts according to it. But as here we have median as a subcolumn of 

median_pivot = df.pivot_table(values=['salary_year_avg'], index='job_title_short' ,  aggfunc=['median', 'min', 'max'])

median_pivot.sort_values(by=('median', 'salary_year_avg'), inplace=True, ascending=False)

median_pivot.sort_index(inplace=True)
# The best use-case of sort_index comes when the column doesn't have a name, Alphabetically sorted here though !

median_pivot
```
