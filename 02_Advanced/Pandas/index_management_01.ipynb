```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
# Index Attributes
# Notes
# You can get or set properties of the index, such as its name or data type.
# It's helpful for maintaining metadata or ensuring index compatibility in operations.
# (index.name, index.dtype)
# index.name - name of the index
# index.dtype - data type of the index

df.sample(5)

df.index # RangeIndex(start=0, stop=785741, step=1)
df.index.dtype # dtype('int64')
df.index.name # NOTHING YET
```

```python
# Setting Index name
df.index.name = 'job_index'
df.index.name
```

```python
df[['job_title_short', 'salary_year_avg']].sample(3) # We say renamed index column as Column -> 'job_index'
```

```python
# Seeing Index names of pivoted tables and printing the list of indices

median_pivot = df.pivot_table(values='salary_year_avg', index='job_title_short', aggfunc='median')
print(median_pivot, end = "\n\n\n") # job_title_short sits down, so it's the index
print(median_pivot.index.name, end = "\n\n\n")
print(median_pivot.index, end = "\n\n\n")
```
