```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Load Data
raw_data = load_dataset("lukebarousse/data_jobs")
df = raw_data['train'].to_pandas()

# Cleanup data
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])

# See the type of each column of the Dataframe
print(type(df['salary_hour_avg']))
```

```python
## BELOW ARE THE THINGS HAPPENING WHEN WE RUN df['column_name'] = df['column_name'] + 2
# Extraction: Pulling the column_name out as a Series.
# Transformation: Creating a new, temporary Series where every number is increased by 2.
# Alignment: Putting that new Series back into the DataFrame. Pandas matches the indices automatically to make sure the right values go into the right rows.

# fillna() is available for both Series and Dataframes
```

```python
# TASK-1: Generate a new Dataframe where the NA values of salary_year_avg and salary_hour_avg is replaced by the 
# median of the available values for those columns

salary_year_avg_median = df['salary_year_avg'].median()
salary_hour_avg_median = df['salary_hour_avg'].median()

filled_df = df

filled_df['salary_year_avg'] = filled_df['salary_year_avg'].fillna(salary_year_avg_median)
filled_df['salary_hour_avg'] = filled_df['salary_hour_avg'].fillna(salary_hour_avg_median)

print(filled_df['salary_year_avg'].head())
print("\n")
print(filled_df['salary_hour_avg'].head())
print("\n")
print(filled_df.info())
```

```python
# Task-2: Removing all Duplicates from the filled_df Dataframe #

# Assign df_unique to filled_df defined in previous Cell
df_unique = filled_df

# Use Dataframe.drop_duplicates(subset=None) to remove all Duplicate Rows from the DF, considering all Column Values
df_unique = df_unique.drop_duplicates()

# Checking how many Rows were actually dropped ---> 0 Rows dropped as exact matching rows for all those Col values were not found
print("---------------------------------------------------------------------------------------\nRemoving all Duplicate Rows with values of all columns same - \n---------------------------------------------------------------------------------------\n")
print("Original filled_df length: ", len(filled_df))
print("New Dataframe with all Duplicates removed length: ", len(df_unique))
print("Number of Rows removed: ", len(filled_df)-len(df_unique))




print("\n\n")



# Providing a subset, and only to remove those Rows which have duplicate values in the subset column only
df_unique = filled_df

# Drop the repetitive Jobs from the Same companies
df_unique = df_unique.drop_duplicates(subset = ['job_title', 'company_name'])

# Checking how many Rows were actually dropped ---> 0 Rows dropped as exact matching rows for all those Col values were not found
print("---------------------------------------------------------------------------------------\nRemoving all Duplicate Rows with values of job_title_short and company_name as same - \n---------------------------------------------------------------------------------------\n")
print("Original filled_df length: ", len(filled_df))
print("New Dataframe with all Duplicates removed length: ", len(df_unique))
print("Number of Rows removed: ", len(filled_df)-len(df_unique))

 
```
