```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Load Data
raw_data = load_dataset("lukebarousse/data_jobs")
df = raw_data['train'].to_pandas()

# Cleanup data
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
# Use the .sample(n) method to select specific sample(s) from the data
print(df.sample())

# Provide values of n
print("\n\n\n")
print(df.sample(3)[['job_title_short', 'job_country', 'job_work_from_home']])

# User random_state value to provide the same sample everytime
print("\n\n\n")
print(df.sample(3, random_state=42)[['job_title_short', 'job_country', 'job_work_from_home']])

# Along column axis
print("\n\n\n")
print(df.sample(2, axis=1, random_state=42)) # 1st 2 columns selected, with all Rows.

# Non-consecutive selection of rows or cols not possible, because the goal is randomization, not selection
```

```python
# Using .copy() to deep copy a Dataframe

df2 = df.copy() # Ignore this line and see rest

df_shallow_copied = df2
print("Making changes to Shallow copy and printing - \n\n")
df_shallow_copied['salary_year_avg'] = df_shallow_copied['salary_year_avg'].median()
print("Copied Version - \n")
print(df_shallow_copied['salary_year_avg'].head(3))
print("Original Version also changed - \n")
print(df2['salary_year_avg'].head(3))

df_deep_copied = df.copy(deep=True) # be default deep = True
print("Making changes to Shallow copy and printing - \n\n")
df_deep_copied['salary_year_avg'] = df_deep_copied['salary_year_avg'].median()
print("Copied Version - \n")
print(df_deep_copied['salary_year_avg'].head(3))
print("Original Version remains unchanged - \n")
print(df['salary_year_avg'].head(3))
```
