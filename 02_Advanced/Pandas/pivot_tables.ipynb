### Pivot Tables in Pandas -> Used to Summarize and Aggregate Data inside a Dataframe

```python
# Concept of Pivot Tables:- A Pivot Table is essentially a tool that turns "long" data into "wide" data. It allows you to take a messy list of records and summarize them into a clean, 2-dimensional grid.
# Think of it as a three-way intersection: you pick one column to be your Rows, another to be your Columns, and a third to be the Values (the numbers in the middle).

# #pivot_df = df.pivot_table(
#     values='salary_year_avg',   # What numbers are we calculating?
#     index='job_title_short',    # What goes on the left (Rows)?
#     columns='job_country',      # What goes on the top (Columns)?
#     aggfunc='mean'              # How do we combine data? (mean, median, sum, count)
```

```python
# Main Usage of Pivot tables come when we have multiple entries for a Single Intersection of index
# and column. For example, On the Same Date, a City can have multiple values of Temp and Humidity.
# If we use simple pivoting here, Pandas doesn't know which entry to choose between the 2, 
# and throws a "Index contains duplicate entries, cannot reshape" Error. 

# But if we use pivot_table() here, Pandas will know with the help of our Aggregate function, that
# what should be done with those multiple values to aggregate them into a single value.


# Use pivot_table whenever:
# 1. There is "Noise": You have multiple sensors or multiple timestamps for the same category.

# 2. You want a Summary: You don't care about the individual 10 AM and 4 PM readings; you just want the Daily Average or the Daily Maximum.

# 3. Data is messy: Real-world data almost always has duplicates. Using pivot_table is "defensive coding"—it prevents your app from crashing if a duplicate row accidentally sneaks into your CSV.
```

```python
# If we don't provide a aggregate function, mean() is used to calculate the avg of multiple values and 
# Fill the value.
import pandas as pd


weather_data = [
    # --- May 1st ---
    {"date": "2025-05-01", "city": "Kolkata", "temp": 30, "humidity": 85},      # Morning
    {"date": "2025-05-01", "city": "Kolkata", "temp": 36, "humidity": 70},      # Afternoon
    {"date": "2025-05-01", "city": "Bankura", "temp": 38, "humidity": 45},
    {"date": "2025-05-01", "city": "Bankura", "temp": 44, "humidity": 28},
    {"date": "2025-05-01", "city": "Bhubaneswar", "temp": 31, "humidity": 75},
    {"date": "2025-05-01", "city": "Bhubaneswar", "temp": 37, "humidity": 68},
    {"date": "2025-05-01", "city": "Patna", "temp": 34, "humidity": 42},
    {"date": "2025-05-01", "city": "Patna", "temp": 41, "humidity": 30},
    {"date": "2025-05-01", "city": "Ranchi", "temp": 28, "humidity": 60},
    {"date": "2025-05-01", "city": "Ranchi", "temp": 33, "humidity": 50},
    {"date": "2025-05-01", "city": "Darjeeling", "temp": 15, "humidity": 90},
    {"date": "2025-05-01", "city": "Darjeeling", "temp": 21, "humidity": 75},
    
    # --- May 2nd ---
    {"date": "2025-05-02", "city": "Kolkata", "temp": 31, "humidity": 82},
    {"date": "2025-05-02", "city": "Kolkata", "temp": 35, "humidity": 72},
    {"date": "2025-05-02", "city": "Bankura", "temp": 39, "humidity": 40},
    {"date": "2025-05-02", "city": "Bankura", "temp": 45, "humidity": 25},
    {"date": "2025-05-02", "city": "Bhubaneswar", "temp": 32, "humidity": 72},
    {"date": "2025-05-02", "city": "Bhubaneswar", "temp": 38, "humidity": 65},
    {"date": "2025-05-02", "city": "Patna", "temp": 35, "humidity": 38},
    {"date": "2025-05-02", "city": "Patna", "temp": 42, "humidity": 28},
    {"date": "2025-05-02", "city": "Ranchi", "temp": 29, "humidity": 55},
    {"date": "2025-05-02", "city": "Ranchi", "temp": 34, "humidity": 45},
    {"date": "2025-05-02", "city": "Darjeeling", "temp": 14, "humidity": 92},
    {"date": "2025-05-02", "city": "Darjeeling", "temp": 20, "humidity": 80}
]

df = pd.DataFrame(weather_data)

# df.pivot(index = "city", columns='date') ---> ERROR !

df.pivot_table(index='city', columns='date') # Average values of multiple Temp and Humidity are calculated.
```

```python
# Providing our own Aggregate Functions --> MEDIAN()
df.pivot_table(index='city', columns='date', aggfunc='median')
```

```python
# Providing our own Aggregate Functions --> MAX()
df.pivot_table(index='city', columns='date', aggfunc='max')
```

```python
# Providing our own Aggregate Functions --> COUNT() ---> How many different values occur for a particular Column for an Intersection Point of City + Date
df.pivot_table(index='city', columns='date', aggfunc='count')
```

```python
# Providing our own Aggregate Functions --> COUNT() and also wanting to see only Temp column
# ---> Aggregate(here COUNT()) the different values of "temp" to a single value, for every unique (City + Date) 
# intersection point.
df.pivot_table(index='city', columns='date', values='temp', aggfunc='count')
```

```python
# When you set margins=True, Pandas adds an extra row at the bottom and an extra column on the far  
# right that calculates the Total (or Average, depending on your aggfunc) for that entire row or column.

# 2. What the "All" cells represent
# ➡️ The Bottom Row: Shows the average temperature of all cities combined for each specific date.
# ➡️ The Right Column: Shows the average temperature of all days combined for each specific city.
# ➡️ The Bottom-Right Corner: Shows the "Grand Average"—the mean temperature of the entire dataset.

print(df.pivot_table(index='city', columns='date', values='temp', aggfunc='count', margins=True, margins_name='Total Count'))

df.pivot_table(index='city', columns='date', values=['temp', 'humidity'], aggfunc='mean', margins=True, margins_name='Regional Average')
```

```python
# Using Multiple Aggregate Functions
df.pivot_table(index='city', columns='date', values = 'temp', aggfunc=['mean', 'median'], margins=True, margins_name='Grand Statistics') 
# Multiple margin_names not allowed. Something like 'Grand Statistics' works fine ! 
```
