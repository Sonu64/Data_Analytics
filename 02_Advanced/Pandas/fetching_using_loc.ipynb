```python
#### iloc Rewind ####

import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Load Data
raw_data = load_dataset("lukebarousse/data_jobs")
df = raw_data['train'].to_pandas()

# Cleanup data
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])

print(df.iloc[0:10, 0:2]) # 1st 10 Rows, then 1st 2 columns
```

```python
### Using .loc(), to select Column names by there labels, even for Non-Consecutive Rows or Cols

print("Select the 'salary_yearly_avg' and 'job_work_from_home' columns from the 1st 5 rows, Dropping the NA values of salary_year_avg - \n")
print(df.loc[0:, ['salary_year_avg', 'job_work_from_home']].dropna(subset = ['salary_year_avg']))
print("\n\n\n\n\n")



print("Select the same, but from selected Rows only, 1st, 90th and 561st Row - \n")
print(df.loc[[0, 89, 560], ['salary_year_avg', 'job_work_from_home', 'job_country']])
print("\n\n\n\n\n")



print("Select only those Rows from the dataset where job_country is India and then select the 'job_title_short', 'job_work_from_home' and 'salary_year_avg', remove NA values - \n")
print(df.loc[df['job_country']=="India", ['job_title_short', 'job_country', 'job_posted_date', 'job_work_from_home', 'salary_year_avg']].dropna(subset = 'salary_year_avg').head(10))
```
